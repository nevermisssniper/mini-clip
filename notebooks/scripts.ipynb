```python
import numpy as np

# Get the list of unique image filenames
all_images = sorted(df_clean['image'].unique())
print(f"Total unique images: {len(all_images)}")

# Shuffle deterministically — same split every time you run this
rng = np.random.default_rng(seed=42)
shuffled = rng.permutation(all_images)

# Split sizes
n_train = 6000
n_val = 1000
n_test = len(shuffled) - n_train - n_val  # remainder

train_imgs = shuffled[:n_train]
val_imgs = shuffled[n_train:n_train + n_val]
test_imgs = shuffled[n_train + n_val:]

print(f"Train: {len(train_imgs)} images")
print(f"Val:   {len(val_imgs)} images")
print(f"Test:  {len(test_imgs)} images")

# Sanity check — no overlap
assert len(set(train_imgs) & set(val_imgs)) == 0
assert len(set(train_imgs) & set(test_imgs)) == 0
assert len(set(val_imgs) & set(test_imgs)) == 0
print("No overlap between splits ✓")

# Save to disk
from pathlib import Path
splits_dir = Path("../data/splits")
splits_dir.mkdir(exist_ok=True)

for name, imgs in [("train", train_imgs), ("val", val_imgs), ("test", test_imgs)]:
    with open(splits_dir / f"{name}.txt", "w") as f:
        f.write("\n".join(imgs))
    print(f"Wrote {splits_dir / f'{name}.txt'}")

# Expected output:
# Total unique images: 8091
# Train: 6000 images
# Val:   1000 images
# Test:  1091 images
# No overlap between splits ✓
# Wrote ../data/splits/train.txt
# Wrote ../data/splits/val.txt
# Wrote ../data/splits/test.txt
#
# Why seed=42: any fixed integer works. 42 is convention. The point
# is that re-running this cell gives the same split every time.
# If you ever want to "rebalance" by trying a different split,
# change the seed — never just re-shuffle without recording it.
```
